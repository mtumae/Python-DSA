**Question 1**: Write a program that asks the user for a number of days. The program then
prints out the number of seconds in the amount of days given.

````python
print("This program will output the number of seconds in the amount of days given.")
while True:
    try:
        prompt = int(input("Enter a number of days: "))
        print(f"{prompt} days is quivalent to {prompt * 24 * 60 * 60:.2f}s.")
    except ValueError:
        print("Invalid input. Try again")
````

**Question 2**: Write a program that asks a user to input the radius then the program
calculates the volume of a sphere (the formula for the volume is (4/3)πr3.)
Use the exponential operator in python to compute (r3)

````python
print("This program will calculate the volume of a sphere given a radius.")
while True:
    try:
        radius = int(input("Enter the radius: "))
        print(f"{(4/3)*3.14*radius**3:.2f}")
    except ValueError:
        print("Invalid input. Try again")
````

**Question 3**: Using functions, write a program to compute the area and perimeter of a
square. The program should ask the user to enter a number corresponding to
the side length of the square and display the area and perimeter of the
square.

````python
print("This program can compute area and perimeter of a square given the side length.")
def area(side_len:int):
    return side_len**2

def perimeter(side_len:int):
    return side_len*4

while True:
    try:
        side_len = int(input("Enter the length of the side of the square: "))
        print(f"For a square of sides {side_len}, \narea:{area(side_len)}, perimeter:{perimeter(side_len)}")
    except ValueError:
        print("Invalid input! Try again.")


````

**Question 4**: Write a program using functions that determines whether a character input
by a user is uppercase or lower case.

````python

print("This program determines whether a character input is uppercase or lower case.")

while True:
    try:
        char = str(input("Enter a single character: "))
        if ord(char) == ord(char.upper()):
            print("The character you input is uppercase.")
        else:
            print("The character you input is lowercase.")
    except ValueError:
        print("Invalid input. Try again.")

````

**Question 5**: The following is pseudocode for a program being designed. Write the
Python program equivalent of the same.

```
    BEGIN
    SET x TO 0, y TO 20
    REPEAT
    SUBTRACT 4 FROM y
    ADD 2/y TO x
    UNTIL
    y IS LESS THAN 6
    DISPLAY x
    END
```


````python
x = 0
y = 20

while y > 6:
    y-=4
    x+=2/y
    
print(f"{x:.2f}")
````

**Question 6**: Write a Python program that does the following:
• Uses a loop for a user to continually input 5 values to populate an
array.
• Calculates and displays the average of the values input into the array.


````python
import random
arr = []
count = 0

for i in range(5):
    arr.append(random.randint(0,9))

for i in arr:
    count +=i


print(f"Average of {arr} is {count/5:.2f}")
````
